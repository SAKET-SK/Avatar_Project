```python
import os
import io
import time
import requests
import numpy as np
from pathlib import Path
from dotenv import load_dotenv

from openai import OpenAI

from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_community.vectorstores import FAISS
from langchain_community.document_loaders import (
    PyPDFLoader,
    TextLoader,
    UnstructuredWordDocumentLoader,
    UnstructuredPowerPointLoader,
    UnstructuredExcelLoader,
)
from langchain_core.prompts import ChatPromptTemplate
# Updated imports for newer versions of langchain
from langchain.chains.retrieval import create_retrieval_chain
from langchain.chains.combine_documents.stuff import create_stuff_documents_chain

import pygame
import scipy.io.wavfile as wavfile

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
DID_API_KEY    = os.getenv("DID_API_KEY", "")

LLM_MODEL     = "gpt-4o"
EMBED_MODEL   = "text-embedding-3-small"
WHISPER_MODEL = "whisper-1"
TTS_MODEL     = "tts-1"
TTS_VOICE     = "nova"   # alloy | echo | fable | onyx | nova | shimmer

# Improved assertion check - only checks if API key exists and is not empty
assert OPENAI_API_KEY and OPENAI_API_KEY.strip() != "", \
    "ERROR: OPENAI_API_KEY not set. Edit .env and add your key."

# Additional check to ensure it's not a placeholder key (starts with 'sk-' for real OpenAI keys)
assert OPENAI_API_KEY.startswith("sk-"), \
    "ERROR: Invalid OPENAI_API_KEY format. Make sure you're using a real OpenAI API key."

client = OpenAI(api_key=OPENAI_API_KEY)

pygame.mixer.init()

print("OpenAI client initialized.")
print(f"LLM model       : {LLM_MODEL}")
print(f"Embedding model : {EMBED_MODEL}")
print(f"TTS voice       : {TTS_VOICE}")
print(f"D-ID key        : {'found' if DID_API_KEY else 'not set (avatar video disabled)'}")
```

### Configuring System Prompt

```python
PHARMA_SYSTEM_PROMPT = """
You are Dr. Sam, a pharmaceutical AI assistant with expertise in:
- Drug interactions and contraindications
- Medication storage and handling
- Dosage guidelines and administration routes
- Regulatory and compliance information
- Adverse effect profiles

IMPORTANT RULES:
1. Always base answers on the retrieved context provided. If context is
   insufficient, say so clearly.
2. For any clinical or patient-specific question, always recommend consulting
   a licensed pharmacist or physician.
3. Keep responses concise and suitable for spoken delivery — no bullet points,
   no markdown, no lists. Write as if you are speaking naturally to someone.
4. If asked something outside pharmaceutical topics, politely redirect.
5. Be precise with drug names — use both generic and brand names where relevant.

Your tone: professional, clear, empathetic, safety-first.
""".strip()

print("System prompt configured.")
print(f"Character count: {len(PHARMA_SYSTEM_PROMPT)}")
```

```python
DOCS_DIR = Path("knowledge_docs")
DOCS_DIR.mkdir(exist_ok=True)

VECTOR_STORE_PATH = "pharma_vectorstore"

LOADER_MAP = {
    ".pdf":  PyPDFLoader,
    ".txt":  TextLoader,
    ".docx": UnstructuredWordDocumentLoader,
    ".doc":  UnstructuredWordDocumentLoader,
    ".pptx": UnstructuredPowerPointLoader,
    ".ppt":  UnstructuredPowerPointLoader,
    ".xlsx": UnstructuredExcelLoader,
    ".xls":  UnstructuredExcelLoader,
}

SUPPORTED_EXTENSIONS = list(LOADER_MAP.keys())


def load_all_documents(docs_dir: Path) -> list:
    """
    Walk knowledge_docs/ and load every supported file.
    Returns a flat list of LangChain Document objects.
    """
    all_docs = []
    found_files = []

    for ext in SUPPORTED_EXTENSIONS:
        found_files.extend(docs_dir.glob(f"*{ext}"))
        found_files.extend(docs_dir.glob(f"*{ext.upper()}"))

    found_files = list(set(found_files))

    if not found_files:
        print("No supported files found in knowledge_docs/")
        print(f"Supported types: {', '.join(SUPPORTED_EXTENSIONS)}")
        return []

    for file_path in sorted(found_files):
        ext = file_path.suffix.lower()
        loader_class = LOADER_MAP.get(ext)

        if loader_class is None:
            print(f"  Skipping (unsupported): {file_path.name}")
            continue

        try:
            if loader_class == TextLoader:
                loader = loader_class(str(file_path), encoding="utf-8")
            else:
                loader = loader_class(str(file_path))

            docs = loader.load()
            all_docs.extend(docs)
            print(f"  Loaded [{ext.upper():5s}]: {file_path.name}  →  {len(docs)} section(s)")

        except Exception as e:
            print(f"  ERROR loading {file_path.name}: {e}")

    return all_docs


# Check if knowledge_docs has any files
existing_files = []
for ext in SUPPORTED_EXTENSIONS:
    existing_files.extend(DOCS_DIR.glob(f"*{ext}"))

if not existing_files:
    print("knowledge_docs/ folder is empty.")
    print("Creating a sample file so the pipeline can run for demo...\n")

    sample_text = """PHARMACEUTICAL STORAGE GUIDELINES

Temperature Sensitivity:
Most medications should be stored at controlled room temperature between 15 and
25 degrees Celsius. Insulin and many biologics require refrigeration between 2
and 8 degrees Celsius. Frozen vaccines must be maintained at minus 15 to minus
25 degrees Celsius. Medications exposed to temperatures outside their specified
range may lose potency or become unsafe.

Light and Humidity:
Many drugs are photosensitive and must be stored in amber or opaque containers
away from direct sunlight. Humidity above 60 percent can cause tablets to
dissolve, clump, or degrade prematurely. Bathroom cabinets are generally
unsuitable for drug storage due to humidity from showers.

Common Drug Interactions:
Warfarin (Coumadin) has numerous clinically significant interactions. NSAIDs
such as ibuprofen increase bleeding risk when combined with warfarin. Antibiotics
including ciprofloxacin and metronidazole can potentiate warfarin's anticoagulant
effect. INR monitoring is essential when starting or stopping any medication in
a warfarin patient.

Serotonin Syndrome Risk:
Combining two or more serotonergic agents can cause serotonin syndrome, a
potentially life-threatening condition. High-risk combinations include SSRIs with
MAOIs, tramadol, or triptans. Symptoms include hyperthermia, agitation, clonus,
and autonomic instability.

Renal Dosage Adjustment:
Many drugs require dose reduction in patients with impaired renal function.
Creatinine clearance or eGFR is used to calculate appropriate dose adjustments.
Metformin is contraindicated when eGFR falls below 30 mL per minute per 1.73 m
squared.
"""
    (DOCS_DIR / "sample_pharma_guide.txt").write_text(sample_text)
    print("Created: knowledge_docs/sample_pharma_guide.txt")

print(f"\nLoading documents from: {DOCS_DIR.resolve()}")
all_docs = load_all_documents(DOCS_DIR)
print(f"\nTotal sections loaded: {len(all_docs)}")

if not all_docs:
    raise RuntimeError("No documents loaded. Add files to knowledge_docs/ and re-run this cell.")

# Split into overlapping chunks
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
    separators=["\n\n", "\n", ". ", " "],
)
chunks = splitter.split_documents(all_docs)
print(f"Total chunks after splitting: {len(chunks)}")

# Embeddings
embeddings = OpenAIEmbeddings(
    model=EMBED_MODEL,
    openai_api_key=OPENAI_API_KEY
)

# Build or load FAISS vector store
if Path(VECTOR_STORE_PATH).exists():
    print("\nLoading existing vector store from disk...")
    vector_store = FAISS.load_local(
        VECTOR_STORE_PATH,
        embeddings,
        allow_dangerous_deserialization=True
    )
    print("Vector store loaded.")
else:
    print("\nBuilding vector store — calling OpenAI embeddings API...")
    vector_store = FAISS.from_documents(chunks, embeddings)
    vector_store.save_local(VECTOR_STORE_PATH)
    print(f"Vector store saved to: {VECTOR_STORE_PATH}/")

retriever = vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 3}
)

print("\nRAG retriever ready.")
```

```python
RAG_SYSTEM_PROMPT = """You are Dr. Sam, a pharmaceutical AI assistant.
Use ONLY the following retrieved context to answer the question.
If the context does not contain enough information, say clearly that you do not
have sufficient information in your knowledge base and recommend consulting a
pharmacist.

Respond in spoken, natural language. No bullet points, no markdown formatting.
Keep the response under 150 words so it is suitable for voice output.

{context}"""

rag_prompt = ChatPromptTemplate.from_messages([
    ("system", RAG_SYSTEM_PROMPT),
    ("human", "{input}"),
])

llm = ChatOpenAI(
    model=LLM_MODEL,
    temperature=0.2,
    openai_api_key=OPENAI_API_KEY,
    max_tokens=300,
)

question_answer_chain = create_stuff_documents_chain(llm, rag_prompt)
qa_chain = create_retrieval_chain(retriever, question_answer_chain)

print("RAG QA chain initialized.")
print(f"LLM: {LLM_MODEL}  |  Temperature: 0.2  |  Max tokens: 300")
```

```python
def ask_pharma_question(question: str, verbose: bool = True) -> str:
    result = qa_chain.invoke({"input": question})
    answer = result["answer"]
    sources = result.get("context", [])

    if verbose:
        print(f"Question: {question}")
        print("-" * 60)
        print(f"Answer:\n{answer}")
        print()
        if sources:
            print("Sources used:")
            for i, doc in enumerate(sources, 1):
                source_name = doc.metadata.get("source", "unknown")
                preview = doc.page_content[:80].replace("\n", " ")
                print(f"  [{i}] {source_name}: '{preview}...'")

    return answer


test_questions = [
    "What temperature should insulin be stored at?",
    "Can I take ibuprofen if I am on warfarin?",
    "How should I dose medications in patients with kidney disease?",
]

for q in test_questions:
    print("=" * 70)
    ask_pharma_question(q)
    print()
```

### Text to Speech layer

```python
def text_to_speech(
    text: str,
    voice: str = TTS_VOICE,
    output_path: str = "avatar_response.mp3",
    play_audio: bool = True
) -> str:
    """
    Convert text to speech using OpenAI TTS API.
    Returns path to the generated audio file.
    """
    print(f"Generating speech ({voice} voice)...")

    response = client.audio.speech.create(
        model=TTS_MODEL,
        voice=voice,
        input=text,
        speed=0.95,
    )

    response.stream_to_file(output_path)
    print(f"Audio saved to: {output_path}")

    if play_audio:
        pygame.mixer.music.load(output_path)
        pygame.mixer.music.play()
        while pygame.mixer.music.get_busy():
            time.sleep(0.1)
        print("Playback complete.")

    return output_path


# Test TTS
test_text = (
    "Good morning. I am Dr. Sam, your pharmaceutical assistant. "
    "Insulin must be stored in a refrigerator between two and eight degrees "
    "Celsius until it is opened. Once opened, most insulin vials can be kept "
    "at room temperature, below 25 degrees, for up to 28 days. Always check "
    "the specific product label and consult your pharmacist for guidance."
)

audio_file = text_to_speech(test_text, voice="nova")
print(f"\nTTS pipeline verified. File: {audio_file}")
```

### Voice Input Layer

```python
try:
    import sounddevice as sd
    AUDIO_AVAILABLE = True
except ImportError:
    AUDIO_AVAILABLE = False
    print("sounddevice not available — text input fallback will be used.")


def record_audio(duration_seconds: int = 5, sample_rate: int = 16000) -> bytes:
    """Record from microphone. Returns raw WAV bytes."""
    if not AUDIO_AVAILABLE:
        raise RuntimeError("sounddevice not available")

    print(f"Recording for {duration_seconds} seconds... Speak now!")

    audio_data = sd.rec(
        int(duration_seconds * sample_rate),
        samplerate=sample_rate,
        channels=1,
        dtype="int16"
    )
    sd.wait()
    print("Recording complete.")

    wav_buffer = io.BytesIO()
    wavfile.write(wav_buffer, sample_rate, audio_data)
    return wav_buffer.getvalue()


def transcribe_audio(wav_bytes: bytes) -> str:
    """Transcribe WAV bytes with OpenAI Whisper. Returns transcript string."""
    audio_file = io.BytesIO(wav_bytes)
    audio_file.name = "recording.wav"

    print("Transcribing with Whisper...")

    transcript = client.audio.transcriptions.create(
        model=WHISPER_MODEL,
        file=audio_file,
        language="en",
        response_format="text"
    )

    print(f"Transcript: '{transcript}'")
    return transcript.strip()


def get_user_question(duration_seconds: int = 5) -> str:
    """Get a question via mic if available, typed text fallback otherwise."""
    if AUDIO_AVAILABLE:
        try:
            wav_bytes = record_audio(duration_seconds)
            return transcribe_audio(wav_bytes)
        except Exception as e:
            print(f"Microphone error: {e} — falling back to text input.")

    return input("Type your question: ")


print("Voice input layer ready.")
print(f"Microphone: {'enabled' if AUDIO_AVAILABLE else 'not available — text mode'}")
```

### D-ID Avatar Video Layer

```python
def generate_avatar_video(
    text: str,
    avatar_image_path: str = "avatar.jpg",
    voice: str = "en-US-JennyNeural",
    output_path: str = "avatar_output.mp4"
) -> str:
    """
    Generate a talking avatar video using the D-ID API.
    Returns path to downloaded MP4, or None if unavailable.
    """
    DID_API_URL = "https://api.d-id.com"

    if not DID_API_KEY:
        print("D-ID API key not set — skipping avatar video.")
        print("Add DID_API_KEY to your .env to enable this.")
        return None

    if not Path(avatar_image_path).exists():
        print(f"Avatar image not found: {avatar_image_path}")
        print("Place your photo as 'avatar.jpg' in the project folder.")
        return None

    headers = {
        "Authorization": f"Basic {DID_API_KEY}",
        "Content-Type": "application/json",
        "Accept": "application/json",
    }

    # Step 1 — Upload avatar image
    print("Uploading avatar image to D-ID...")
    with open(avatar_image_path, "rb") as img_file:
        upload_response = requests.post(
            f"{DID_API_URL}/images",
            headers={"Authorization": f"Basic {DID_API_KEY}"},
            files={"image": (avatar_image_path, img_file, "image/jpeg")}
        )
    upload_response.raise_for_status()
    image_url = upload_response.json()["url"]
    print(f"Image uploaded: {image_url}")

    # Step 2 — Request talking video
    print("Requesting video generation...")
    talk_payload = {
        "script": {
            "type": "text",
            "input": text,
            "provider": {
                "type": "microsoft",
                "voice_id": voice
            }
        },
        "source_url": image_url,
        "config": {
            "stitch": True,
            "result_format": "mp4"
        }
    }

    create_response = requests.post(
        f"{DID_API_URL}/talks",
        headers=headers,
        json=talk_payload
    )
    create_response.raise_for_status()

    talk_id = create_response.json()["id"]
    print(f"Talk ID: {talk_id} — polling for completion...")

    # Step 3 — Poll until done
    max_wait = 60
    interval = 3
    elapsed = 0

    while elapsed < max_wait:
        time.sleep(interval)
        elapsed += interval

        status_response = requests.get(
            f"{DID_API_URL}/talks/{talk_id}",
            headers=headers
        )
        status_data = status_response.json()
        status = status_data.get("status")
        print(f"  Status: {status} ({elapsed}s)")

        if status == "done":
            video_bytes = requests.get(status_data["result_url"]).content
            with open(output_path, "wb") as f:
                f.write(video_bytes)
            print(f"Avatar video saved: {output_path}")
            return output_path

        elif status == "error":
            print(f"D-ID error: {status_data.get('error', 'unknown')}")
            return None

    print(f"Timeout: video not ready after {max_wait}s.")
    return None


print("D-ID avatar layer configured.")
print(f"Status: {'ACTIVE' if DID_API_KEY else 'INACTIVE — no DID_API_KEY set'}")
```

### Full Pipeline Orchestrator

```python
def run_avatar_pipeline(
    use_voice_input: bool = True,
    recording_duration: int = 5,
    generate_video: bool = False,
    tts_voice: str = TTS_VOICE
):
    print("\n" + "=" * 70)
    print("AI PHARMA AVATAR — Pipeline starting")
    print("=" * 70)

    # STEP 1 — Get question
    print("\n[STEP 1] Input")
    if use_voice_input and AUDIO_AVAILABLE:
        question = get_user_question(duration_seconds=recording_duration)
    else:
        question = input("Your question: ")

    if not question.strip():
        print("No question detected. Stopping.")
        return

    print(f"Question: {question}")

    # STEP 2 — RAG
    print("\n[STEP 2] RAG — retrieving knowledge and generating answer")
    t0 = time.time()

    result = qa_chain.invoke({"input": question})
    answer = result["answer"]
    sources = result.get("context", [])

    print(f"Answer generated in {time.time() - t0:.1f}s")
    print(f"\nAnswer:\n{answer}")

    if sources:
        source_names = [doc.metadata.get("source", "?") for doc in sources]
        print(f"\nSources used: {source_names}")

    # STEP 3 — TTS
    print("\n[STEP 3] Text-to-Speech")
    t1 = time.time()

    audio_path = text_to_speech(
        answer,
        voice=tts_voice,
        output_path="avatar_response.mp3",
        play_audio=True
    )

    print(f"TTS completed in {time.time() - t1:.1f}s")

    # STEP 4 — Avatar video (optional)
    if generate_video:
        print("\n[STEP 4] Avatar Video (D-ID)")
        video_path = generate_avatar_video(
            text=answer,
            output_path="avatar_output.mp4"
        )
        if video_path:
            print(f"Video saved: {video_path}")

    print("\n" + "=" * 70)
    print("Pipeline complete.")
    print("=" * 70)

    return {
        "question": question,
        "answer": answer,
        "audio_path": audio_path,
        "sources": [doc.metadata.get("source", "?") for doc in sources]
    }


print("Pipeline orchestrator ready.")
```

```python
import time, requests
from pathlib import Path

def generate_video_with_presenter(
    text: str,
    presenter_id: str = PRESENTER_ID,
    voice_id: str = PRESENTER_VOICE,
    voice_type: str = PRESENTER_VOICE_TYPE,
    output_path: str = "avatar_output.mp4"
) -> str:
    """
    Generate a talking avatar video using a D-ID built-in presenter.
    Uses the correct presenter_id key and voice structure from the API.
    """
    DID_API_URL = "https://api.d-id.com"

    if not DID_API_KEY or DID_API_KEY == "your_did_api_key_here":
        print("D-ID API key not set — skipping video generation.")
        return None

    headers = {
        "Authorization": f"Basic {DID_API_KEY}",
        "Content-Type": "application/json",
        "Accept": "application/json",
    }

    payload = {
        "presenter_id": presenter_id,
        "script": {
            "type": "text",
            "input": text,
            "provider": {
                "type": voice_type,       # "microsoft" or "elevenlabs"
                "voice_id": voice_id
            }
        },
        "config": {
            "result_format": "mp4"
        }
    }

    print(f"Requesting video...")
    print(f"Presenter : {presenter_id}")
    print(f"Voice     : {voice_type} — {voice_id}")

    try:
        response = requests.post(
            f"{DID_API_URL}/clips",
            headers=headers,
            json=payload
        )

        if response.status_code != 201:
            print(f"Error {response.status_code}: {response.text}")
            return None

    except Exception as e:
        print(f"Request failed: {e}")
        return None

    clip_id = response.json()["id"]
    print(f"Clip ID   : {clip_id}")
    print(f"Processing...")

    max_wait = 120
    interval = 4
    elapsed  = 0

    while elapsed < max_wait:
        time.sleep(interval)
        elapsed += interval

        status_response = requests.get(
            f"{DID_API_URL}/clips/{clip_id}",
            headers=headers
        )
        status_data = status_response.json()
        status = status_data.get("status")
        print(f"  [{elapsed}s] Status: {status}")

        if status == "done":
            video_url   = status_data["result_url"]
            video_bytes = requests.get(video_url).content
            with open(output_path, "wb") as f:
                f.write(video_bytes)
            size_kb = len(video_bytes) / 1024
            print(f"\nVideo saved: {output_path} ({size_kb:.1f} KB)")
            return output_path

        elif status == "error":
            print(f"D-ID error: {status_data.get('error', 'unknown')}")
            return None

    print(f"Timeout after {max_wait}s.")
    return None


print("Video generation function ready.")
```
